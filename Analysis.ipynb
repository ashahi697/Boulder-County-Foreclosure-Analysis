```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

df = pd.read_csv("data/analysis_data.csv",
                 parse_dates=["sale_date"],
                 low_memory=False
                )

print("Total records:", len(df))
display(df.head())
    
```

```python
year_summary = (
    df.groupby("sale_year")
    .agg(
        number_of_sales=("sale_price", "count"),
        median_sale_price=("sale_price", "median"),
        average_sale_price=("sale_price", "mean")
    )
    .reset_index()
)

display(year_summary)
```

```python
fig, ax = plt.subplots(figsize=(9, 5))

ax.bar(
    year_summary["sale_year"].astype(str),
    year_summary["median_sale_price"]
)

ax.set_title("Median Property Sale Price in Boulder County")
ax.set_xlabel("Sale Year")
ax.set_ylabel("Median Sale Price ($)")

fig.tight_layout()

Path("figures").mkdir(exist_ok=True)
fig.savefig("figures/median_sale_price.png", dpi=300)

plt.show()
```
